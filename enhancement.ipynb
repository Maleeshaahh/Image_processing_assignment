```python
import cv2
import os
import numpy as np

# =========================
# INPUT ROOT FOLDER
# =========================
input_folder = "Frame_extraction"

# =========================
# OUTPUT ROOT FOLDERS
# =========================
gaussian_root = "gaussian_output"
clahe_root = "clahe_output"
sharpen_root = "final_enhanced"

os.makedirs(gaussian_root, exist_ok=True)
os.makedirs(clahe_root, exist_ok=True)
os.makedirs(sharpen_root, exist_ok=True)

# =========================
# CLAHE SETUP
# =========================
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# =========================
# SHARPENING KERNEL
# =========================
kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

# =========================
# PROCESS ALL SUBFOLDERS
# =========================
print("Starting pipeline...\n")

for root, dirs, files in os.walk(input_folder):

    print(f"Scanning folder: {root}")

    for filename in files:

        # only images
        if not filename.lower().endswith((".jpg", ".png", ".jpeg")):
            continue

        image_path = os.path.join(root, filename)
        frame = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

        if frame is None:
            print("Cannot read:", image_path)
            continue

        # get match folder name (match1, match2, etc.)
        match_name = os.path.basename(root)

        # =========================
        # CREATE OUTPUT SUBFOLDERS
        # =========================
        gaussian_folder = os.path.join(gaussian_root, match_name)
        clahe_folder = os.path.join(clahe_root, match_name)
        sharpen_folder = os.path.join(sharpen_root, match_name)

        os.makedirs(gaussian_folder, exist_ok=True)
        os.makedirs(clahe_folder, exist_ok=True)
        os.makedirs(sharpen_folder, exist_ok=True)

        # =========================
        # STEP 1 - GAUSSIAN BLUR
        # =========================
        gaussian = cv2.GaussianBlur(frame, (3, 3), 0)
        gaussian_path = os.path.join(gaussian_folder, f"gaussian_{filename}")
        cv2.imwrite(gaussian_path, gaussian)

        # =========================
        # STEP 2 - CLAHE
        # =========================
        clahe_image = clahe.apply(gaussian)
        clahe_path = os.path.join(clahe_folder, f"clahe_{filename}")
        cv2.imwrite(clahe_path, clahe_image)

        # =========================
        # STEP 3 - SHARPENING
        # =========================
        sharpened = cv2.filter2D(clahe_image, -1, kernel)
        sharpen_path = os.path.join(sharpen_folder, f"enhanced_{filename}")
        cv2.imwrite(sharpen_path, sharpened)

        print(f"Processed: {image_path}")

print("\n✅ Enhancement pipeline completed successfully!")
```
