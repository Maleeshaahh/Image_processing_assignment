```python
import cv2
import os
import numpy as np

input_root = "final_enhanced"
output_root = "segmented_frames"

os.makedirs(output_root, exist_ok=True)

kernel = np.ones((3, 3), np.uint8)

for match_folder in sorted(os.listdir(input_root)):

    match_input_path = os.path.join(input_root, match_folder)

    if not os.path.isdir(match_input_path):
        continue

    match_output_path = os.path.join(output_root, match_folder)
    os.makedirs(match_output_path, exist_ok=True)

    print(f"Processing: {match_folder}")

    # reset for each match
    bg_subtractor = cv2.createBackgroundSubtractorMOG2(
        history=300,
        varThreshold=40,
        detectShadows=False
    )

    for file in sorted(os.listdir(match_input_path)):

        frame_path = os.path.join(match_input_path, file)
        frame = cv2.imread(frame_path)

        if frame is None:
            continue

        # optional: mild blur on COLOR frame
        blur = cv2.GaussianBlur(frame, (5, 5), 0)

        # foreground mask
        mask = bg_subtractor.apply(blur)

        # cleanup
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.dilate(mask, kernel, iterations=2)

        # save
        output_path = os.path.join(match_output_path, file)
        cv2.imwrite(output_path, mask)

print("Segmentation completed for all matches!")
```
